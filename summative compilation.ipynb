# Imagining China: Cross-National Attitudes Towards Information Freedom

## 1. Introduction

The *People’s Republic of China* occupies a prominent position in global debates on information freedom, frequently portrayed as the best example of a digital authoritarian state in many liberal democracies [1]. Drawing on data from the 2025 Pew Global Attitudes Survey, this report compares citizen-level opinions from 35 countries. It investigates whether cross-national perceptions of China correlate with broader attitudes towards information governance and freedom, with a particular emphasis placed on attitudes towards internet, speech, and press freedom. Where systematic patterns emerge, it treats perceptions of China as a lens onto citizens’ own value commitments around freedom and control, and asks what these patterns may reveal about wider global power relations, particularly between global majority and non-majority nations.

## 2. Literature Review

### 2.1 Theorising Cross-National Perceptions: From Social Psychology to Critical Theory

Cross-national survey research has long shown that people in different countries place varying degrees of importance on freedom and self-expression, suggesting that a country’s value profile may also shape how its own citizens evaluate foreign states, and what forms of information control they deem as acceptable [2]. Social identity theory suggests that individuals derive part of their own self-concept from the groups they belong to, defining these ‘in-groups’ in contrast to relevant ‘out-groups’ [3]. International relations scholars have extended this logic to the analysis of states, emphasising how nations construct an “Other” against which to articulate their own core values and anxieties [4]. Critical and reflexive approaches, often associated with the “self-reflexive turn” in anthropology [5, 6], brought questions of positionality, assumptions, and complicity in power relations to the forefront of social science research, prompting scholars to consider how narratives about others may reflect their own fears, interests, and contradictions [7]. Edward Said’s 1978 publication of Orientalism famously epitomised this shift by theorising how a Western imaginary of “the East” legitimises hierarchies between Self and Other [8]. Subsequent work has traced how such dynamics are reproduced in digital contexts, including forms of “techno-orientalism” in contemporary representations of Asian societies [9, 10, 11, 12]. Together, these perspectives suggest that perceptions of China can be read not only as judgements about an external actor, but also as reflections of the societies doing the observing.

## 2.2 Digital Freedom and Perceptions of China

Despite having over 1.12 billion internet users and an internet penetration rate of around 79.7% as of 2025 [13], China is consistently ranked among the most repressive countries for information freedom, receiving the lowest score for political rights and civil liberties from Freedom House [14] and placing 178th out of 180 countries in the Reporters Without Borders 2025 World Press Freedom Index [15]. As highlighted in much Western media, the Chinese Communist Party retains extensive control over digital infrastructure [16, 17], online speech [18], and information flows [19] through bespoke content moderation systems and internet censorship mechanisms, such as China’s “Great Firewall” [20, 21, 22]. Simultaneously, amidst intensifying competition with the United States, China’s pioneering advances in technology place it at the centre of global innovation [23]. China is an economic superpower, the world’s largest trading nation and the second-largest economy by nominal GDP [24]. This colossal weight creates ambivalent relationships with many nations, who publicly condemn China’s digital policies while remaining deeply entwined with Chinese trade and investment. For example, despite maintaining robust trading ties, the United Kingdom and the United States consider China to be one of their most pressing security and economic threats [25, 26].

However, public attitudes towards China are far from uniform across world regions. Much existing research and media commentary foregrounds perspectives from Euro-American liberal democracies, while the views of populations in the “global majority” remains comparatively underexamined [27, 28]. "Global majority" is a term used for people residing in Asia, Africa, Latin America and the Middle East, who constitute 85% of the global population. The term has also been used as an alternative to language which embeds eurocentric perspectives normatively, for example, uses of the term "Third World Country". China’s global image is complex and contested [29]. National perceptions of China are not simply neutral, but also mirror domestic debates about freedom, security, and technology governance. Although some associations made about China within liberal democratic discourse may be true, the gaze through which such opinions are formed deserve closer scrutiny. For example, this relationship raises questions about how governmental and media depictions of distant states shape citizens’ attitudes. Whilst much political rhetoric focuses on contrasting liberal democracies with authoritarian regimes, both China and the United States exhibit significant convergences when regulating the internet, deploying emerging technologies, and confronting tensions around privacy, surveillance, content moderation and information control [30].


## Methodology

### 3.1 Research Question and Hypothesis

How do cross-national perceptions of China correlate with attitudes toward information freedom, and what do these patterns reveal about positionality and power relations between nations? It is expected that stronger support for internet, speech and press freedom will be associated with more negative perceptions of China. Furthermore, it is hypothesised that this relationship will be more true for Western liberal democracies, whereas attitudes in many global-majority regions will be more mixed.

### 3.2 Data 

This analysis uses the Pew Spring Global Attitudes 2025 Survey, which collected data from a sample of n = 41,503 participants across 35 countries. The population of interest (Ω) consists of all participants from these 35 countries and the dataset used contains 654 variables. This study focuses on variables measuring favourability towards China, measured on a 4-point ordinal scale: "very favourable" (1), "somewhat favourable" (2), "somewhat unfavourable" (3), and "very unfavourable" (4). Variables measuring national importance placed on internet freedom, press freedom, and free speech are also examined. These are similarly measured using a 4-point ordinal scale, where respondents indicated whether each freedom was "very important" (1), "somewhat important" (2), "not too important" (3), or "not important at all" (4). Numeric labels (1–4) encode these scales, with lower values indicating more positive attitudes. All variables examined in this report were administered to all participants across the 35 countries. Variables vary in their total number of valid responses. Non-response codes 8 ("don't know") and 9 ("refused") are also present in the dataset, but are excluded during analysis and considered non-valid following Pew's methodological guidance. For example, information freedom variables have between 1,951 – 2,204 missing or non-valid responses (4.7% – 5.3% of n), which are dropped from the analysis. Where made explicit, country-level statistics are calculated as weighted proportions also following Pew guidance, to ensure balanced representativeness of demographics during cross-national comparisons. 

```python
import plotly as pl
import matplotlib.pyplot as plt
import seaborn as s
import matplotlib as mp
import numpy as np
import pandas as pd
import os
import pathlib
```

```python
pew = pd.read_csv("pew_dataset.csv", dtype=str) 
```

```python
# convert relevant columns to numeric

pew['weight'] = pd.to_numeric(pew['weight'], errors='coerce')
pew['age'] = pd.to_numeric(pew['age'], errors='coerce')  
pew['id'] = pd.to_numeric(pew['id'], errors='coerce')  
pew['country'] = pd.to_numeric(pew['country'], errors='coerce') 

# create a python dictionary, generating an additional 'country' column from country-codes for easier readability

country_code_map = {
    2: "Canada",
    3: "France",
    4: "Germany",
    5: "Greece",
    6: "Hungary",
    7: "Italy",
    8: "Netherlands",
    9: "Poland",
    10: "Spain",
    11: "Sweden",
    12: "UK",
    13: "Australia",
    14: "Bangladesh",
    15: "India",
    16: "Indonesia",
    17: "Japan",
    18: "Malaysia",
    19: "Philippines",
    20: "Singapore",
    21: "South Korea",
    22: "Sri Lanka",
    23: "Thailand",
    24: "Israel",
    25: "Tunisia",
    26: "Turkey",
    27: "Ghana",
    28: "Kenya",
    29: "Nigeria",
    30: "South Africa",
    31: "Argentina",
    32: "Brazil",
    33: "Chile",
    34: "Colombia",
    35: "Mexico",
    36: "Peru"
}

pew['country_name'] = pew['country'].map(country_code_map)
```

```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.colors as mcolors

# Filter to valid responses
valid_data = pew[pew['fav_china'].isin(['1', '2', '3', '4'])].copy()
valid_data['weight'] = pd.to_numeric(valid_data['weight'], errors='coerce')
valid_data = valid_data.dropna(subset=['weight', 'country_name', 'fav_china'])

# Calculate weighted proportions for each country
country_stats = {}
for country in sorted(valid_data['country_name'].unique()):
    country_data = valid_data[valid_data['country_name'] == country]
    total_weight = country_data['weight'].sum()
    
    fav_china_favorable = country_data[country_data['fav_china'].isin(['1', '2'])]['weight'].sum()
    fav_china_favorable_pct = (fav_china_favorable / total_weight) * 100
    
    country_stats[country] = {
        'fav_china_favorable_pct': fav_china_favorable_pct
    }

plot_df = pd.DataFrame(country_stats).T

# Sort by China favorability for better visualization
plot_df = plot_df.sort_values('fav_china_favorable_pct', ascending=True)

# Create colormap for favorability gradient (low to high)
# Using RdYlGn: Red (low favorability) -> Yellow -> Green (high favorability)
cmap = plt.cm.RdYlGn  # Red to Green
norm = mcolors.Normalize(vmin=plot_df['fav_china_favorable_pct'].min(), 
                        vmax=plot_df['fav_china_favorable_pct'].max())

# Create lollipop chart with favorability gradient coloring
fig, ax = plt.subplots(figsize=(12, max(10, len(plot_df) * 0.4)))

y_pos = np.arange(len(plot_df))

# Plot lollipop with favorability gradient colors
for i, country in enumerate(plot_df.index):
    china_val = plot_df.loc[country, 'fav_china_favorable_pct']
    color = cmap(norm(china_val))  # Get color from gradient based on favorability
    
    # Draw line
    ax.plot([0, china_val], [i, i], color=color, linewidth=2, alpha=0.7, zorder=1)
    
    # Draw dot with gradient color
    ax.scatter(china_val, i, s=200, color=color, alpha=0.8, edgecolors='black', linewidth=1.5, zorder=3)
    
    # Add percentage text to the right of the dot
    ax.text(china_val + 2, i, f'{china_val:.1f}%', 
            va='center', fontsize=9, fontweight='bold', zorder=4)

# Create mini world map inset with favorability gradient dots
ax_map = fig.add_axes([0.55, 0.55, 0.40, 0.45], projection=ccrs.PlateCarree())

# Add base geographic features
ax_map.add_feature(cfeature.COASTLINE, linewidth=0.5, color='gray')
ax_map.add_feature(cfeature.BORDERS, linewidth=0.3, color='gray')
ax_map.add_feature(cfeature.OCEAN, alpha=0.5)
ax_map.add_feature(cfeature.LAND, alpha=0.2)

# Country coordinates for map (approximate centroids)
country_coords = {
    'UK': (0, 54), 'France': (2, 46), 'Germany': (10, 51), 'Italy': (12, 42),
    'Spain': (-3, 40), 'Poland': (20, 52), 'Netherlands': (5, 52), 'Sweden': (18, 60),
    'Greece': (23, 39), 'Hungary': (20, 47),
    'Canada': (-95, 56), 'United States': (-95, 38),
    'Japan': (138, 36), 'South Korea': (127, 36), 'India': (77, 20), 'Indonesia': (120, -5),
    'Malaysia': (102, 4), 'Philippines': (121, 13), 'Singapore': (104, 1), 'Thailand': (101, 15),
    'Bangladesh': (90, 24), 'Sri Lanka': (81, 7),
    'Israel': (35, 31), 'Tunisia': (10, 34), 'Turkey': (35, 39),
    'South Africa': (25, -29), 'Ghana': (-1, 8), 'Kenya': (38, 0), 'Nigeria': (8, 9),
    'Brazil': (-55, -15), 'Argentina': (-64, -34), 'Chile': (-71, -33), 'Colombia': (-74, 4),
    'Peru': (-75, -10), 'Mexico': (-102, 23),
    'Australia': (133, -25)
}

# Plot countries as dots with favorability gradient colors
for country in plot_df.index:
    if country in country_coords:
        lon, lat = country_coords[country]
        favorability = plot_df.loc[country, 'fav_china_favorable_pct']
        color = cmap(norm(favorability))  # Get color from gradient
        
        ax_map.scatter(lon, lat, s=80, c=[favorability], cmap=cmap, 
                      vmin=norm.vmin, vmax=norm.vmax, alpha=0.9, 
                      edgecolors='black', linewidth=1.5, zorder=5, 
                      transform=ccrs.PlateCarree())

# Add colorbar for favorability gradient
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax_map, orientation='horizontal', 
                   pad=0.05, shrink=0.6, aspect=20)
cbar.set_label('Favourability (%)', fontsize=8, labelpad=5)
cbar.ax.tick_params(labelsize=7)

# Set map bounds
ax_map.set_global()
ax_map.set_extent([-180, 180, -70, 90], crs=ccrs.PlateCarree())
ax_map.set_title('Geographic Distribution', fontsize=9, fontweight='bold', pad=5)

# Main plot settings
ax.set_yticks(y_pos)
ax.set_yticklabels(plot_df.index, fontsize=10)
ax.set_xlabel('Favourability of China (Weighted Relative Frequencies of Positive Responses, %)', fontsize=12)
ax.set_ylabel('Country', fontsize=12)
ax.set_title('Fig. 1: Global Perceptions of China, Country-Level Favourability',
             fontsize=14, fontweight='bold', pad=20)
ax.set_xlim(0, 105)  # Extended slightly to accommodate percentage text
ax.set_xticks(np.arange(0, 101, 20))
ax.grid(True, alpha=0.3, axis='x', linestyle='--')
ax.invert_yaxis()  # Top to bottom

plt.show()
```

An initial analysis measured favourability toward China (`fav_china`) across all 35 countries to establish a baseline distribution. Upon excluding 4,232 non-valid or missing responses (10.2% of n = 41,503), the final sample included 37,271 valid responses. Expressed as relative frequencies for each country $c$, favourability was operationalised as the weighted proportion $p_c$ of participants reporting "very favourable" (1) or "somewhat favorable" (2). Figure 1 demonstrated that baseline China favourability varied significantly across nations, ranging from 11.5% to 82.2%, with a median of 50.8%. The analysis illustrated that Sweden (11.5%), Japan (11.7%) and Australia (14.0%) showed the lowest favourability, while Nigeria (81.7%), Thailand (81.2%) and Bangladesh (77.1%) exhibited the highest. As visualised by the 'geographic distribution' graphic in Fig. 1, countries typically associated with the Global Majority demonstrated systematically higher favorability toward China, with all countries ≥ 52.9% being Global Majority nations, with the exception of Tunisia (73.1%). Western nations and Euro-American liberal democracies clustered in the first two quartiles, with most in Q1 (median favourability of 26%). India (25.6%) and the Philippines (34.8%) were the only Global Majority countries scoring below the overall median (< 50.8%).

## Analysis


```python
# Clean Combined Visualization: China Favorability vs All Three Freedom Variables
# Using subplots for clarity

# Reuse calculation function from before or recalculate
def calc_china_freedom_correlation(freedom_var):
    valid_data = pew[
        pew['fav_china'].isin(['1', '2', '3', '4']) & 
        pew[freedom_var].isin(['1', '2', '3', '4'])
    ].copy()
    valid_data['weight'] = pd.to_numeric(valid_data['weight'], errors='coerce')
    valid_data = valid_data.dropna(subset=['weight', 'country_name', 'fav_china', freedom_var])
    
    country_stats = {}
    for country in sorted(valid_data['country_name'].unique()):
        country_data = valid_data[valid_data['country_name'] == country]
        total_weight = country_data['weight'].sum()
        
        fav_china_favorable = country_data[country_data['fav_china'].isin(['1', '2'])]['weight'].sum()
        fav_china_favorable_pct = (fav_china_favorable / total_weight) * 100
        
        freedom_important = country_data[country_data[freedom_var].isin(['1', '2'])]['weight'].sum()
        freedom_important_pct = (freedom_important / total_weight) * 100
        
        country_stats[country] = {
            'fav_china_favorable_pct': fav_china_favorable_pct,
            'freedom_important_pct': freedom_important_pct
        }
    
    return pd.DataFrame(country_stats).T

plot_df_internet = calc_china_freedom_correlation('internet_freedom')
plot_df_speech = calc_china_freedom_correlation('free_speech')
plot_df_press = calc_china_freedom_correlation('press_freedom')

# Create figure with 3 subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

from scipy import stats
from scipy.stats import spearmanr

# Define colors for each
colors = ['#e74c3c', '#3498db', '#2ecc71']
datasets = [
    (plot_df_internet, 'Internet Freedom', 'internet_freedom'),
    (plot_df_speech, 'Free Speech', 'free_speech'),
    (plot_df_press, 'Press Freedom', 'press_freedom')
]

for idx, (plot_df, title, var_name) in enumerate(datasets):
    ax = axes[idx]
    
    # Scatter plot
    ax.scatter(plot_df['fav_china_favorable_pct'], 
               plot_df['freedom_important_pct'], 
               s=100, alpha=0.7, color=colors[idx], edgecolors='black', linewidth=0.5)
    
    # Trend line
    z = np.polyfit(plot_df['fav_china_favorable_pct'], 
                   plot_df['freedom_important_pct'], 1)
    p = np.poly1d(z)
    ax.plot(plot_df['fav_china_favorable_pct'], 
            p(plot_df['fav_china_favorable_pct']), 
            color='black', linestyle='--', alpha=0.6, linewidth=2)
    
    # Calculate correlation
    corr, p_val = spearmanr(plot_df['fav_china_favorable_pct'], 
                            plot_df['freedom_important_pct'])
    
    # Customize
    ax.set_xlabel('China Favorability (%)', fontsize=11)
    if idx == 0:
        ax.set_ylabel('Freedom Importance (%)', fontsize=11)
    ax.set_title(f'{title}\nr = {corr:.3f}, p = {p_val:.4f}', 
                 fontsize=12, fontweight='bold')
    ax.grid(True, alpha=0.2, linestyle='--')
    ax.set_xlim(0, 100)
    ax.set_ylim(0, 100)

plt.suptitle('Fig. 2: China Favourability and Information Freedom', 
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

Correlations between favourability towards China and importance placed on information freedoms were analysed. For each country $c$, ordinal responses were converted into weighted proportions $p_c$ representing the percentage of participants rating each freedom as "very" or "somewhat" important (values 1 or 2). Spearman's rank correlation coefficient $ρ$ was used to assess relationships between favourability towards China and each freedom variable. Spearman's correlation was chosen due to its appropriateness for working with ordinal data, and its capacity to measure monotonic relationships by ranking observations $ω$ rather than assuming linearity, making it robust to the possibility of non-normal distributions. The p-value $p$ represents the probability that a correlation of magnitude $|r|$ or greater would occur by chance if no true relationship existed between the variables (null hypothesis $\mathrm{H}_0: ρ = 0$). Figure 2 presents scatterplots of this relationship, with correlation coefficients r and p-values $p$ reported. For internet freedom, $r = -0.533$, $p = 0.0012$; for free speech, $r = -0.60$, $p = 0.0001$; and for press freedom, $r = -0.554$, $p = 0.0007$. P-values ranging from 0.0001 to 0.0012 indicate that the observed correlations would occur by chance less than 0.12% of the time if no true relationship existed, providing strong evidence against $\mathrm{H}_0$. Strong negative correlations were observed across all three freedom variables, indicating that countries with higher favorability toward China tend to place lower importance on internet, speech, and press freedoms, consistent with the hypothesis.
