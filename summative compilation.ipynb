# Imagining China: Cross-National Attitudes Towards Internet Freedom and Information Governance 

## 1. Introduction

The *People’s Republic of China* occupies a prominent position in global debates on information freedom, frequently portrayed as the best example of a digital authoritarian state in many liberal democracies [1]. Drawing on data from the 2025 Pew Global Attitudes Survey, this report compares citizen-level opinions from 35 countries. It investigates whether cross-national perceptions of China correlate with broader attitudes towards information governance, with a particular emphasis placed on attitudes towards internet and press freedom. Where systematic patterns emerge, it treats perceptions of China as a lens onto citizens’ own value commitments around freedom and control, and asks what these patterns may reveal about wider global power relations, particularly between global majority and non-majority nations.

## 2. Literature Review

### 2.1 Theorising Cross-National Perceptions: From Social Psychology to Critical Theory

Cross-national survey research has long shown that people in different countries place varying degrees of importance on freedom and self-expression, suggesting that a country’s value profile may also shape how its own citizens evaluate foreign states, and what forms of information control they deem as acceptable [2]. Social identity theory suggests that individuals derive part of their own self-concept from the groups they belong to, defining these ‘in-groups’ in contrast to relevant ‘out-groups’ [3]. International relations scholars have extended this logic to the analysis of states, emphasising how nations construct an “Other” against which to articulate their own core values and anxieties [4]. Critical and reflexive approaches, often associated with the “self-reflexive turn” in anthropology [5, 6], brought questions of positionality, assumptions, and complicity in power relations to the forefront of social science research, prompting scholars to consider how narratives about others may reflect their own fears, interests, and contradictions [7]. Edward Said’s 1978 publication of Orientalism famously epitomised this shift by theorising how a Western imaginary of “the East” legitimises hierarchies between Self and Other [8]. Subsequent work has traced how such dynamics are reproduced in digital contexts, including forms of “techno-orientalism” in contemporary representations of Asian societies [9, 10, 11, 12]. Together, these perspectives suggest that perceptions of China can be read not only as judgements about an external actor, but also as reflections of the societies doing the observing.

## 2.2 Digital Freedom and Perceptions of China

Despite having over 1.12 billion internet users and an internet penetration rate of around 79.7% as of 2025 [13], China is consistently ranked among the most repressive countries for information freedom, receiving the lowest score for political rights and civil liberties from Freedom House [14] and placing 178th out of 180 countries in the Reporters Without Borders 2025 World Press Freedom Index [15]. As highlighted in much Western media, the Chinese Communist Party retains extensive control over digital infrastructure [16, 17], online speech [18], and information flows [19] through bespoke content moderation systems and internet censorship mechanisms, such as China’s “Great Firewall” [20, 21, 22]. Simultaneously, amidst intensifying competition with the United States, China’s pioneering advances in technology place it at the centre of global innovation [23]. China is an economic superpower, the world’s largest trading nation and the second-largest economy by nominal GDP [24]. This colossal weight creates ambivalent relationships with many nations, who publicly condemn China’s digital policies while remaining deeply entwined with Chinese trade and investment. For example, despite maintaining robust trading ties, the United Kingdom and the United States consider China to be one of their most pressing security and economic threats [25, 26].

However, public attitudes towards China are far from uniform across world regions. Much existing research and media commentary foregrounds perspectives from Euro-American liberal democracies, while the views of populations in the “global majority” [27, 28], broadly referring to vast populations in Asia, Africa, Latin America and the Middle East – remain comparatively underexamined. China’s global image is complex and contested [29]. National perceptions of China are not simply neutral, but also mirror domestic debates about freedom, security, and technology governance. Although some associations made about China within liberal democratic discourse may be true, the gaze through which such opinions are formed deserve closer scrutiny. This relationship raises questions about how governmental and media depictions of distant states shape citizens’ attitudes. Whilst much political rhetoric focuses on contrasting liberal democracies with authoritarian regimes, both China and the United States exhibit significant convergences with regards to regulating the internet, deploying new technologies, and confronting social tensions around privacy, surveillance, content moderation and information control [30].
